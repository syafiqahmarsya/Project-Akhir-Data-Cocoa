# Uji Normalitas Model Terbaik

Akan dilakukan uji normalitas pada model terbaik yang terpilih (model hasil metode stepwise (model_asumsi.step))

## Standarisasi Residual

```r
## Standarisasi Residual
resid1 <-residuals(model_asumsi.step)
resid2 <-rstandard(model_asumsi.step)
```

## Hipotesis

 Asumsikan Uji hipotesisnormalitas:
* H0: residual mengikutidistribusi normal
* H1: residual tidak mengikutidistribusi normal

## Uji Shapiro Wilk

```r
shapiro.test(resid1)
```

berdasarkan uji shapiro-wilk didapat bahwa p-value < 0.05 , sehingga tolak H0 dan residual model tersebut tidak mengikuti distribusi normal

 ## uji kolmogorov-smirnov

```r
ks.test(resid2, "pnorm", mean = 0, sd= 1)
```

berdasarkan uji kolmogorov-smirnov didapat bahwa p-value < 0.05 , sehingga tolak H0 dan residual model tersebut tidak mengikuti distribusi normal



```r
ols_test_normality(model_asumsi.step)
```

## Grafik distribusi residual model

```r
# Melihat bentuk grafik (histogram dan plot) untuk melihat apakah garfik menunjukan residual berdidtribusi normal atau tidak
hist(resid2, xlab= "residual", main = "")
ols_plot_resid_qq(model_asumsi.step)
```

berdasarkan grafik terlihat bahwa residual model tidak berdistribusi normal sehingga dilakukan transformasi variabel respon


# Transformasi Model

```r
# Transformasi
bc<-boxcox(cocoa$hv ~ blk + wilt + lrg + smi + pcp + smi:pcp, data = cocoa)
lambda<-bc$x[which.max(bc$y)]
print(lambda)
if(lambda==0){
  hv_trans<-log(cocoa$hv)
} else{
  hv_trans<-(cocoa$hv^lambda-1)/lambda
}
cocoa$hv_trans<-hv_trans

```

```r
#model dengan data hasil transformasi

model_final <- lm(formula = hv_trans~ blk + wilt + lrg + smi + pcp + smi:pcp, data = cocoa)
summary(model_final)

```

 dilakukan uji normalitas kembali untuk melihat apakah residual model sudah berdistribusi normal
